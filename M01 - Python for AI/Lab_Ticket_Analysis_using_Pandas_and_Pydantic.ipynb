## Step 9 – Create the Pydantic Ticket Model

### Explanation

Pandas helps us read, organise, and analyse data, but it does not automatically enforce the business rules of our project.

For example, a ticket should not have:

- A ticket ID of zero or less.
- A blank customer name.
- An unsupported category or priority.
- A status other than `Open` or `Closed`.
- A negative resolution time.

**Pydantic** checks whether data matches the rules we define.

We will create a `Ticket` model. Each field in this model represents one piece of ticket information, and the validation rules ensure that the data follows our project requirements.

### Rules for This Lab

| Field | Validation rule |
|---|---|
| `ticket_id` | Must be greater than zero |
| `customer_name` | Must not be blank |
| `category` | Must be `Technical`, `Billing`, or `Account` |
| `priority` | Must be `Low`, `Medium`, or `High` |
| `status` | Must be `Open` or `Closed` |
| `resolution_hours` | Must be zero or greater |

### Your Task

1. Create a new code cell below this text cell.
2. Copy the code provided below into the code cell.
3. Run the cell to create the `src/models.py` file.
4. Check the output to confirm that the file has been created successfully.

We will use the `%%writefile` command at the beginning of the code cell. It saves the code into the specified Python file instead of executing it directly in the notebook.



````python
%%writefile src/models.py
from typing import Literal

from pydantic import BaseModel, Field, field_validator


class Ticket(BaseModel):
    ticket_id: int = Field(gt=0)
    customer_name: str = Field(min_length=1)
    category: Literal["Technical", "Billing", "Account"]
    priority: Literal["Low", "Medium", "High"]
    status: Literal["Open", "Closed"]
    resolution_hours: float = Field(ge=0)

    @field_validator("customer_name")
    @classmethod
    def customer_name_must_not_be_blank(cls, value):
        value = value.strip()
        if not value:
            raise ValueError("Customer name cannot be blank")
        return value
````



### Expected Result

---

### Understanding the Model

Let's understand the important parts of the code:

- **`BaseModel`** – The Pydantic class that provides data validation.
- **`Field(gt=0)`** – Ensures that the value is greater than zero. We use this for `ticket_id`.
- **`Field(ge=0)`** – Ensures that the value is greater than or equal to zero. We use this for `resolution_hours`.
- **`Literal[...]`** – Restricts a field to only the specified options. For example, `priority` can only be `Low`, `Medium`, or `High`.
- **`field_validator`** – Allows us to define a custom validation rule for a field.
- **`@classmethod`** – Defines the custom validator as a class method.
- **`strip()`** – Removes extra spaces from the beginning and end of a customer name.
- **`ValueError`** – Raises an error if the customer name is empty or contains only spaces.

**Important:** `Field(min_length=1)` checks that the name contains at least one character. The custom validator adds another check to reject names containing only spaces.

**Great!** We have created our Pydantic `Ticket` model and defined the validation rules for our project. In the next step, we will use this model to validate ticket records.

## Step 11 – Write a Function to Validate All CSV Rows

### Explanation

In the previous steps, we created a Pydantic model and tested it with individual ticket records.

Now we will connect **Pandas and Pydantic** to validate all the records in a CSV file.

We will create a Python file named `analyzer.py` inside the `src/` folder. This file will contain two functions:

1. **`load_and_validate_tickets()`** – Reads a CSV file, checks each ticket using Pydantic, and separates valid and invalid records.
2. **`analyze_tickets()`** – Analyses the valid tickets and calculates ticket counts and the average resolution time for closed tickets.

Our validation function will follow these steps:

1. Read a CSV file into a Pandas DataFrame.
2. Visit each row one at a time.
3. Convert each row into a dictionary.
4. Ask Pydantic to validate the record.
5. Store valid records and invalid records in separate lists.
6. Return the valid records as a DataFrame and the invalid records as a list.

We use `iterrows()` to go through the DataFrame one row at a time. We use `model_validate()` to check each record and `model_dump()` to convert a validated Pydantic model back into a dictionary.

### Your Task

1. Create a new code cell below this text cell.
2. Copy the code provided below into the code cell.
3. Run the cell to create the `src/analyzer.py` file.
4. Check the output to confirm that the file has been created successfully.

We will use the `%%writefile` command to save the code into the Python file.



````python
%%writefile src/analyzer.py
import pandas as pd
from pydantic import ValidationError

from src.models import Ticket


def load_and_validate_tickets(csv_path):
    dataframe = pd.read_csv(csv_path)

    valid_records = []
    invalid_records = []

    for row_number, row in dataframe.iterrows():
        record = row.to_dict()

        try:
            ticket = Ticket.model_validate(record)
            valid_records.append(ticket.model_dump())
        except ValidationError as error:
            invalid_records.append({
                "row_number": int(row_number) + 2,
                "record": record,
                "error": str(error)
            })

    valid_df = pd.DataFrame(valid_records)
    return valid_df, invalid_records


def analyze_tickets(valid_df):
    if valid_df.empty:
        return {
            "total_tickets": 0,
            "category_counts": {},
            "priority_counts": {},
            "status_counts": {},
            "average_closed_resolution_hours": None
        }

    closed_tickets = valid_df[valid_df["status"] == "Closed"]

    average_resolution = None
    if not closed_tickets.empty:
        average_resolution = round(
            closed_tickets["resolution_hours"].mean(), 2
        )

    return {
        "total_tickets": int(len(valid_df)),
        "category_counts": valid_df["category"].value_counts().to_dict(),
        "priority_counts": valid_df["priority"].value_counts().to_dict(),
        "status_counts": valid_df["status"].value_counts().to_dict(),
        "average_closed_resolution_hours": average_resolution
    }
````

## Step 15 – Create Automated Tests Using Pytest

### Explanation

Testing helps us check whether our code works correctly and produces the expected results.

In this project, we will use **Pytest**, a Python testing framework, to test the functions we created in `src/analyzer.py`.

Pytest automatically identifies functions whose names start with `test_` and runs them.

We use the Python `assert` statement to compare the actual result with the expected result.

For example:

```python
assert len(valid_df) == 10
```

This checks whether the DataFrame contains exactly 10 valid records. If the condition is true, the test passes. If it is false, the test fails.

````python
%%writefile tests/test_analyzer.py
import pandas as pd

from src.analyzer import load_and_validate_tickets, analyze_tickets


def test_valid_csv_has_ten_valid_records():
    valid_df, invalid_records = load_and_validate_tickets("data/tickets.csv")
    assert len(valid_df) == 10
    assert len(invalid_records) == 0


def test_mixed_csv_separates_valid_and_invalid_records():
    valid_df, invalid_records = load_and_validate_tickets(
        "data/tickets_with_invalid_records.csv"
    )
    assert len(valid_df) == 10
    assert len(invalid_records) == 5


def test_total_ticket_count():
    valid_df, _ = load_and_validate_tickets("data/tickets.csv")
    summary = analyze_tickets(valid_df)
    assert summary["total_tickets"] == 10


def test_category_counts():
    valid_df, _ = load_and_validate_tickets("data/tickets.csv")
    summary = analyze_tickets(valid_df)
    assert summary["category_counts"] == {
        "Technical": 4, "Billing": 3, "Account": 3
    }


def test_priority_counts():
    valid_df, _ = load_and_validate_tickets("data/tickets.csv")
    summary = analyze_tickets(valid_df)
    assert summary["priority_counts"] == {
        "High": 4, "Medium": 4, "Low": 2
    }


def test_status_counts():
    valid_df, _ = load_and_validate_tickets("data/tickets.csv")
    summary = analyze_tickets(valid_df)
    assert summary["status_counts"] == {"Closed": 7, "Open": 3}


def test_average_closed_resolution_time():
    valid_df, _ = load_and_validate_tickets("data/tickets.csv")
    summary = analyze_tickets(valid_df)
    assert summary["average_closed_resolution_hours"] == 4.29


def test_empty_dataframe():
    empty_df = pd.DataFrame(columns=[
        "ticket_id", "customer_name", "category",
        "priority", "status", "resolution_hours"
    ])
    summary = analyze_tickets(empty_df)
    assert summary["total_tickets"] == 0
    assert summary["average_closed_resolution_hours"] is None
````
